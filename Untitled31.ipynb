```python
# ════════════════════════════════════════════════════════════
# STEP 1 & 2: YOLOv11 LANDMARK EXTRACTION
# ════════════════════════════════════════════════════════════

import os
import random
import zipfile
import numpy as np
import pandas as pd
from pathlib import Path

# ============================================================
# GLOBAL CONFIGURATION FOR LANDMARK EXTRACTION
# ============================================================

YOLO_LANDMARK_LABELS = [
    "nose", "left_eye", "right_eye", "left_ear", "right_ear",
    "left_shoulder", "right_shoulder", "left_elbow", "right_elbow",
    "left_wrist", "right_wrist", "left_hip", "right_hip",
    "left_knee", "right_knee", "left_ankle", "right_ankle"
]

# Define the landmarks required for the current task
REQUIRED_LANDMARKS = [
    "left_shoulder", "right_shoulder", "left_elbow", "right_elbow",
    "left_wrist", "right_wrist", "left_hip", "right_hip",
    "left_knee", "right_knee", "left_ankle", "right_ankle"
]

CSV_SAVE_PATH = "landmarks.csv"

RANDOM_STATE = 42

# ------------------------------------------------------------
# Extract ZIP automatically if needed
# ------------------------------------------------------------
if not Path(DATASET_ROOT).exists():
    zip_path = "/content/SleepStressDataset_2 (1).zip"  # Change if your zip has another name

    if Path(zip_path).exists():
        print(f"Extracting {zip_path}...")
        with zipfile.ZipFile(zip_path, "r") as zip_ref:
            zip_ref.extractall(DATASET_ROOT)
        print("Dataset extracted successfully.\n")


def load_yolo_model():
    """
    Load YOLOv11 pose model.
    """
    try:
        from ultralytics import YOLO

        print("Loading YOLOv11 pose model...")
        model = YOLO("yolo11x-pose.pt")

        print("YOLOv11 loaded successfully.")
        return model, True

    except ImportError:
        print("[WARNING] ultralytics not installed.")
        print("Install using: pip install ultralytics")
        return None, False

    except Exception as e:
        print(f"[WARNING] Could not load YOLOv11: {e}")
        return None, False


def extract_landmarks_yolo(model, image_path):
    """
    Extract normalized pose landmarks using YOLOv11.
    Returns:
        dict {landmark_name: [x,y]}
        None if no person detected
    """

    try:
        results = model.predict(
            source=str(image_path),
            save=False,
            verbose=False
        )

        if not results:
            return None

        result = results[0]

        if result.boxes is None or len(result.boxes) == 0:
            return None

        if result.keypoints is None:
            return None

        # Select largest detected person
        boxes = result.boxes.xyxy.cpu().numpy()

        areas = []
        for box in boxes:
            x1, y1, x2, y2 = box
            areas.append((x2 - x1) * (y2 - y1))

        person_idx = np.argmax(areas)

        xyxy = result.boxes.xyxy[person_idx].cpu().numpy()
        keypoints = result.keypoints.xy[person_idx].cpu().numpy()

        x_min, y_min, x_max, y_max = xyxy

        width = max(x_max - x_min, 1e-6)
        height = max(y_max - y_min, 1e-6)

        landmark_data = {}

        for idx, label in enumerate(YOLO_LANDMARK_LABELS):

            if label not in REQUIRED_LANDMARKS:
                continue

            if idx >= len(keypoints):
                landmark_data[label] = None
                continue

            kx, ky = keypoints[idx]

            if kx == 0 and ky == 0:
                landmark_data[label] = None
            else:
                nx = 1000 * (kx - x_min) / width
                ny = 1000 * (ky - y_min) / height

                landmark_data[label] = [
                    float(nx),
                    float(ny)
                ]

        return landmark_data

    except Exception:
        return None


def scan_dataset(dataset_root):
    """
    Scan dataset folders recursively.
    """

    print(f"\nScanning Dataset: {dataset_root}")

    samples = []

    for cls in CLASS_NAMES:

        cls_dir = Path(dataset_root) / cls

        if not cls_dir.exists():
            print(f"[WARNING] Missing Folder: {cls}")
            continue

        imgs = []

        for ext in IMG_EXTS:
            imgs.extend(cls_dir.rglob(f"*{ext}"))

        print(f"{cls:<12}: {len(imgs)} images")

        for img in imgs:
            samples.append((str(img), cls))

    print(f"\nTotal Images Found: {len(samples)}")

    return samples


def _dummy_landmarks():
    """
    Random landmark generator when YOLO unavailable.
    """

    return {
        lm: [
            random.uniform(100, 900),
            random.uniform(100, 900)
        ]
        for lm in REQUIRED_LANDMARKS
    }


def _generate_demo_landmark_df():
    """
    Generate synthetic dataset for testing.
    """

    print("\n[DEMO MODE] Generating synthetic landmarks...")

    np.random.seed(RANDOM_STATE)

    rows = []
    samples_per_class = 120

    for cls in CLASS_NAMES:

        for i in range(samples_per_class):

            row = {
                "image_name": f"demo_{cls}_{i}.jpg",
                "label": cls
            }

            base_x = {
                "Foetus": 400,
                "Log": 500,
                "Starfish": 550,
                "Unknown": 450,
                "Yearner": 480
            }.get(cls, 500)

            for lm in REQUIRED_LANDMARKS:
                row[f"{lm}_x"] = base_x + np.random.randn() * 120
                row[f"{lm}_y"] = 500 + np.random.randn() * 150

            rows.append(row)

    return pd.DataFrame(rows)


def run_landmark_extraction(dataset_root):
    """
    Main landmark extraction pipeline.
    """

    print("\n" + "=" * 60)
    print("STEP 1 & 2 : YOLOv11 LANDMARK EXTRACTION")
    print("=" * 60)

    dataset_path = Path(dataset_root)

    if not dataset_path.exists():
        print(f"[ERROR] Dataset not found: {dataset_root}")
        return _generate_demo_landmark_df()

    samples = scan_dataset(dataset_root)

    if len(samples) == 0:
        print("[WARNING] No images found.")
        return _generate_demo_landmark_df()

    model, yolo_ok = load_yolo_model()

    rows = []
    failed = 0

    for i, (img_path, cls_name) in enumerate(samples):

        if i == 0 or (i + 1) % 50 == 0:
            print(f"Processing {i+1}/{len(samples)}")

        if yolo_ok:
            landmarks = extract_landmarks_yolo(
                model,
                img_path
            )
        else:
            landmarks = _dummy_landmarks()

        if landmarks is None:
            failed += 1
            continue

        row = {
            "image_name": Path(img_path).name,
            "label": cls_name
        }

        for lm in REQUIRED_LANDMARKS:

            coords = landmarks.get(lm)

            if coords is None:
                row[f"{lm}_x"] = np.nan
                row[f"{lm}_y"] = np.nan
            else:
                row[f"{lm}_x"] = coords[0]
                row[f"{lm}_y"] = coords[1]

        rows.append(row)

    df = pd.DataFrame(rows)

    df.to_csv(CSV_SAVE_PATH, index=False)

    print("\n" + "=" * 60)
    print(f"Successful Samples : {len(df)}")
    print(f"Failed Samples     : {failed}")
    print(f"Saved CSV          : {CSV_SAVE_PATH}")
    print("=" * 60)

    return df


# ------------------------------------------------------------
# RUN EXTRACTION
# ------------------------------------------------------------

landmark_df = run_landmark_extraction(DATASET_ROOT)

print("\nDataset Shape:", landmark_df.shape)
print("\nFirst 5 Rows:")
print(landmark_df.head())
```

```python
# ════════════════════════════════════════════════════════════
# STEP 3: MISSING VALUE IMPUTATION
# ════════════════════════════════════════════════════════════

def impute_missing_landmarks(df, n_passes=5):
    """
    Impute missing landmark coordinates using Euclidean-distance
    nearest-neighbor matching, followed by mean imputation for
    any remaining missing values.

    Parameters
    ----------
    df : pandas.DataFrame
        Landmark feature dataframe.
    n_passes : int, default=5
        Number of iterative imputation passes.

    Returns
    -------
    pandas.DataFrame
        Dataframe with imputed landmark coordinates.
    """

    print("\n" + "=" * 60)
    print("STEP 3: IMPUTING MISSING LANDMARKS")
    print("=" * 60)

    df = df.copy()

    # Select numeric landmark coordinate columns
    coord_cols = [
        col for col in df.columns
        if col not in ["image_name", "label"]
    ]

    initial_missing = df[coord_cols].isna().sum().sum()
    print(f"Initial Missing Values : {initial_missing}")

    # --------------------------------------------------------
    # Nearest-Neighbor Imputation (Euclidean Distance)
    # --------------------------------------------------------
    for _ in range(n_passes):

        missing_rows = df[df[coord_cols].isna().any(axis=1)].index

        for idx in missing_rows:

            row = df.loc[idx, coord_cols]

            missing_cols = row[row.isna()].index.tolist()

            if len(missing_cols) == 0:
                continue

            available_cols = row[row.notna()].index.tolist()

            if len(available_cols) == 0:
                continue

            # Candidate rows must have values for all missing columns
            candidate_mask = (
                df[missing_cols].notna().all(axis=1)
            )

            candidate_rows = df.loc[candidate_mask, coord_cols]

            # Remove current row
            candidate_rows = candidate_rows.drop(
                index=idx,
                errors="ignore"
            )

            if len(candidate_rows) == 0:
                continue

            # Distance computed only on available features
            reference_vector = row[available_cols].astype(float).values

            candidate_vectors = (
                candidate_rows[available_cols]
                .astype(float)
                .values
            )

            distances = np.sum(
                (candidate_vectors - reference_vector) ** 2,
                axis=1
            )

            nearest_idx = candidate_rows.index[
                np.argmin(distances)
            ]

            # Fill missing values from nearest row
            for col in missing_cols:
                df.loc[idx, col] = df.loc[nearest_idx, col]

    # --------------------------------------------------------
    # Mean Imputation for Remaining Missing Values
    # --------------------------------------------------------
    for col in coord_cols:

        if df[col].isna().any():

            mean_value = df[col].mean()

            if pd.isna(mean_value):
                # If the mean itself is NaN (e.g., all values were missing),
                # fill with 0 or another suitable default
                df[col].fillna(0, inplace=True)
            else:
                df[col].fillna(mean_value, inplace=True)

    final_missing = df[coord_cols].isna().sum().sum()
    print(f"Final Missing Values   : {final_missing}")

    print("=" * 60 + "\n")

    return df


# Call the imputation function
landmark_df = impute_missing_landmarks(landmark_df)

print("Imputed DataFrame Head:")
print(landmark_df.head())
print("\nMissing values after imputation:")
print(landmark_df.isna().sum().sum())
```

```python

# ════════════════════════════════════════════════════════════
# STEP 4: FEATURE EXTRACTION
# ════════════════════════════════════════════════════════════

def get_pt(row, landmark):
    x = row.get(f'{landmark}_x', 0) or 0
    y = row.get(f'{landmark}_y', 0) or 0
    return np.array([float(x), float(y)])

def euclidean(a, b):
    return float(np.linalg.norm(a - b))

def angle_3pts(A, B, C):
    """Angle at vertex B formed by A-B-C."""
    v1 = A - B;  v2 = C - B
    denom = np.linalg.norm(v1) * np.linalg.norm(v2)
    if denom < 1e-8:
        return 0.0
    cos_a = np.dot(v1, v2) / denom
    return float(np.degrees(np.arccos(np.clip(cos_a, -1.0, 1.0))))


def extract_all_features(df):
    """
    Compute the full feature set described in the paper:
      - Distances between key landmarks
      - Key angles (knee, hip, elbow)
      - Horizontal & vertical spread
      - Symmetry between left/right
      - Trigonometric features (sin, cos of key angles)
    """
    print("\n  Extracting features (distances, angles, symmetry, trig)...")
    feature_rows = []

    for idx, row in df.iterrows():
        f = {
            "image_name": row["image_name"],
            "label": row["label"]
        }

        # ── Landmark shorthand ──
        ls  = get_pt(row, 'left_shoulder');   rs  = get_pt(row, 'right_shoulder')
        le  = get_pt(row, 'left_elbow');      re  = get_pt(row, 'right_elbow')
        lw  = get_pt(row, 'left_wrist');      rw  = get_pt(row, 'right_wrist')
        lh  = get_pt(row, 'left_hip');        rh  = get_pt(row, 'right_hip')
        lk  = get_pt(row, 'left_knee');       rk  = get_pt(row, 'right_knee')
        la  = get_pt(row, 'left_ankle');      ra  = get_pt(row, 'right_ankle')

        # ── Distances ──
        f['shoulder_width']          = euclidean(ls, rs)
        f['hip_width']               = euclidean(lh, rh)
        f['left_shoulder_hip_dist']  = euclidean(ls, lh)
        f['right_shoulder_hip_dist'] = euclidean(rs, rh)
        f['left_hip_knee_dist']      = euclidean(lh, lk)
        f['right_hip_knee_dist']     = euclidean(rh, rk)
        f['left_knee_ankle_dist']    = euclidean(lk, la)
        f['right_knee_ankle_dist']   = euclidean(rk, ra)
        f['left_shoulder_knee_dist'] = euclidean(ls, lk)
        f['right_shoulder_knee_dist']= euclidean(rs, rk)
        f['left_elbow_hip_dist']     = euclidean(le, lh)
        f['right_elbow_hip_dist']    = euclidean(re, rh)
        f['left_wrist_hip_dist']     = euclidean(lw, lh)
        f['right_wrist_hip_dist']    = euclidean(rw, rh)
        f['torso_diagonal']          = euclidean(ls, rh)

        # ── Angles ──
        lka = angle_3pts(lh, lk, la)    # left knee
        rka = angle_3pts(rh, rk, ra)    # right knee
        lha = angle_3pts(ls, lh, lk)    # left hip
        rha = angle_3pts(rs, rh, rk)    # right hip
        lea = angle_3pts(ls, le, lw)    # left elbow
        rea = angle_3pts(rs, re, rw)    # right elbow

        f['left_knee_angle']  = lka;  f['right_knee_angle']  = rka
        f['left_hip_angle']   = lha;  f['right_hip_angle']   = rha
        f['left_elbow_angle'] = lea;  f['right_elbow_angle'] = rea

        # ── Spread ──
        all_x = [ls[0], rs[0], lh[0], rh[0], lk[0], rk[0], la[0], ra[0]]
        all_y = [ls[1], rs[1], lh[1], rh[1], lk[1], rk[1], la[1], ra[1]]
        f['horizontal_spread'] = float(max(all_x) - min(all_x))
        f['vertical_spread']   = float(max(all_y) - min(all_y))
        f['spread_ratio']      = f['horizontal_spread'] / max(f['vertical_spread'], 1e-6)

        # ── Symmetry ──
        f['shoulder_symmetry'] = abs(ls[1] - rs[1])
        f['hip_symmetry']      = abs(lh[1] - rh[1])
        f['knee_symmetry']     = abs(lk[1] - rk[1])
        f['ankle_symmetry']    = abs(la[1] - ra[1])
        f['elbow_symmetry']    = abs(le[1] - re[1])

        # ── Trigonometric ──
        for name, val in [('lka', lka), ('rka', rka),
                           ('lha', lha), ('rha', rha),
                           ('lea', lea), ('rea', rea)]:
            rad = math.radians(val)
            f[f'{name}_sin'] = math.sin(rad)
            f[f'{name}_cos'] = math.cos(rad)
        feature_rows.append(f)

    result_df = pd.DataFrame(feature_rows)
    n_feat = len(result_df.columns) - 2 # Exclude image_name and label
    print(f"  Feature extraction complete: {n_feat} features per sample")
    return result_df


# ------------------------------------------------------------
# RUN FEATURE EXTRACTION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("STEP 4: FEATURE EXTRACTION")
print("=" * 60)

feature_df = extract_all_features(landmark_df)

print("\nFeature DataFrame Shape:", feature_df.shape)
print("\nFirst 5 Rows of Feature DataFrame:")
print(feature_df.head())
```

```python
# ============================================================
# Q-LEARNING FEATURE SELECTION
# ============================================================

import numpy as np
import pandas as pd
import random
import matplotlib.pyplot as plt

from collections import defaultdict

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score

# ============================================================
# PREPARE DATA
# ============================================================

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)

# Remove non-feature columns
X = feature_df.drop(columns=['image_name', 'label'])

# Target column
y = feature_df['label']

# Encode labels
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)

# Normalize features
scaler = StandardScaler()
X = scaler.fit_transform(X)

print("Feature Matrix Shape:", X.shape)
print("Labels Shape:", y.shape)


# ============================================================
# Q-LEARNING FEATURE SELECTION AGENT
# ============================================================

class QLearningFeatureSelector:

    def __init__(self,
                 n_features,
                 n_episodes=60,
                 alpha=0.1,
                 gamma=0.9,
                 epsilon=1.0,
                 min_epsilon=0.1):

        self.n_features = n_features
        self.n_episodes = n_episodes
        self.alpha = alpha
        self.gamma = gamma
        self.epsilon = epsilon
        self.min_epsilon = min_epsilon

        self.q_table = defaultdict(lambda: np.zeros(n_features))

        self.best_subset = None
        self.best_reward = -np.inf

        self.reward_history = []

    def _state_key(self, mask):
        return ''.join(map(str, mask.astype(int)))

    def _evaluate(self, X, y, mask):

        if mask.sum() == 0:
            return 0.0

        selected_features = mask.astype(bool)

        X_sub = X[:, selected_features]

        clf = ExtraTreesClassifier(
            n_estimators=50,
            random_state=RANDOM_STATE,
            n_jobs=-1
        )

        cv = StratifiedKFold(
            n_splits=3,
            shuffle=True,
            random_state=RANDOM_STATE
        )

        scores = []

        for train_idx, test_idx in cv.split(X_sub, y):

            X_train = X_sub[train_idx]
            X_test = X_sub[test_idx]

            y_train = y[train_idx]
            y_test = y[test_idx]

            clf.fit(X_train, y_train)

            preds = clf.predict(X_test)

            acc = accuracy_score(y_test, preds)

            scores.append(acc)

        return np.mean(scores)

    def fit(self, X, y):

        print("\nStarting Q-Learning Feature Selection...")
        print(f"Features : {self.n_features}")
        print(f"Episodes : {self.n_episodes}")

        epsilon_decay = (
            (1.0 - self.min_epsilon) /
            max(self.n_episodes, 1)
        )

        for episode in range(self.n_episodes):

            mask = (
                np.random.rand(self.n_features) > 0.5
            ).astype(float)

            if mask.sum() == 0:
                mask[np.random.randint(self.n_features)] = 1

            steps = min(self.n_features, 20)

            episode_reward = 0

            for step in range(steps):

                state = self._state_key(mask)

                # Epsilon-greedy policy
                if random.random() < self.epsilon:
                    action = random.randint(
                        0,
                        self.n_features - 1
                    )
                else:
                    action = np.argmax(
                        self.q_table[state]
                    )

                new_mask = mask.copy()

                # Toggle feature
                new_mask[action] = 1 - new_mask[action]

                if new_mask.sum() == 0:
                    new_mask[action] = 1

                reward = self._evaluate(
                    X,
                    y,
                    new_mask
                )

                next_state = self._state_key(
                    new_mask
                )

                best_next_q = np.max(
                    self.q_table[next_state]
                )

                # Q-learning update
                self.q_table[state][action] += (
                    self.alpha
                    * (
                        reward
                        + self.gamma * best_next_q
                        - self.q_table[state][action]
                    )
                )

                mask = new_mask

                episode_reward = reward

                if reward > self.best_reward:

                    self.best_reward = reward

                    self.best_subset = (
                        mask.copy()
                    )

            self.reward_history.append(
                episode_reward
            )

            self.epsilon = max(
                self.min_epsilon,
                self.epsilon - epsilon_decay
            )

            if (episode + 1) % 10 == 0:

                print(
                    f"Episode {episode+1:3d}/{self.n_episodes}"
                    f" | Best Accuracy = {self.best_reward:.4f}"
                    f" | Epsilon = {self.epsilon:.3f}"
                )

        print("\nTraining Complete")

        print(
            f"Best CV Accuracy: "
            f"{self.best_reward:.4f}"
        )

        print(
            f"Selected Features: "
            f"{int(self.best_subset.sum())}"
            f"/{self.n_features}"
        )

        return self

    def transform(self, X):
        return X[:, self.best_subset.astype(bool)]

    def plot_rewards(self):

        plt.figure(figsize=(10,5))

        plt.plot(
            self.reward_history,
            linewidth=2
        )

        plt.title(
            "Q-Learning Reward History"
        )

        plt.xlabel("Episode")
        plt.ylabel("Reward")

        plt.grid(True)

        plt.show()


# ============================================================
# RUN Q-LEARNING
# ============================================================

selector = QLearningFeatureSelector(
    n_features=X.shape[1],
    n_episodes=60,
    alpha=0.1,
    gamma=0.9,
    epsilon=1.0,
    min_epsilon=0.1
)

selector.fit(X, y)


# ============================================================
# SELECTED FEATURES
# ============================================================

feature_names = feature_df.drop(
    columns=['image_name', 'label']
).columns

selected_features = feature_names[
    selector.best_subset.astype(bool)
]

print("\nSelected Features:")
print(list(selected_features))

print(
    "\nNumber of Selected Features:",
    len(selected_features)
)


# ============================================================
# REDUCED DATASET
# ============================================================

X_selected = selector.transform(X)

print(
    "\nOriginal Shape:",
    X.shape
)

print(
    "Reduced Shape:",
    X_selected.shape
)


# ============================================================
# PLOT REWARDS
# ============================================================

selector.plot_rewards()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import EfficientNetB3
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

IMG_SIZE = 300
BATCH_SIZE = 16

# Dataset
train_ds = tf.keras.preprocessing.image_dataset_from_directory(
    "/content/dataset",
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)

val_ds = tf.keras.preprocessing.image_dataset_from_directory(
    "/content/dataset",
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)

# Strong augmentation
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.2),
    layers.RandomZoom(0.2),
    layers.RandomContrast(0.2),
])

# EfficientNetB3
base_model = EfficientNetB3(
    weights='imagenet',
    include_top=False,
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

base_model.trainable = False

model = models.Sequential([
    data_augmentation,
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.BatchNormalization(),
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(5, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

early_stop = EarlyStopping(
    monitor='val_accuracy',
    patience=5,
    restore_best_weights=True
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=3,
    min_lr=1e-6
)

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=20,
    callbacks=[early_stop, reduce_lr]
)
```

```python
def extract_resnet_feature(img_path):

    img = cv2.imread(img_path)

    img = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2RGB
    )

    img = cv2.resize(
        img,
        (224,224)
    )

    img = np.expand_dims(
        img,
        axis=0
    )

    img = preprocess_input(img)

    feat = feature_extractor.predict(
        img,
        verbose=0
    )

    return feat.flatten()
```

```python
import numpy as np
import pandas as pd

def dist(x1,y1,x2,y2):
    return np.sqrt((x2-x1)**2 + (y2-y1)**2)

features = []

for _, row in landmark_df.iterrows():

    f = {}

    # Shoulders
    f["shoulder_width"] = dist(
        row["x5"], row["y5"],
        row["x6"], row["y6"]
    )

    # Hips
    f["hip_width"] = dist(
        row["x11"], row["y11"],
        row["x12"], row["y12"]
    )

    # Torso
    f["torso_left"] = dist(
        row["x5"], row["y5"],
        row["x11"], row["y11"]
    )

    f["torso_right"] = dist(
        row["x6"], row["y6"],
        row["x12"], row["y12"]
    )

    # Upper Legs
    f["left_thigh"] = dist(
        row["x11"], row["y11"],
        row["x13"], row["y13"]
    )

    f["right_thigh"] = dist(
        row["x12"], row["y12"],
        row["x14"], row["y14"]
    )

    # Lower Legs
    f["left_leg"] = dist(
        row["x13"], row["y13"],
        row["x15"], row["y15"]
    )

    f["right_leg"] = dist(
        row["x14"], row["y14"],
        row["x16"], row["y16"]
    )

    # Arms
    f["left_arm"] = dist(
        row["x5"], row["y5"],
        row["x7"], row["y7"]
    )

    f["right_arm"] = dist(
        row["x6"], row["y6"],
        row["x8"], row["y8"]
    )

    # Body Height
    f["body_height"] = (
        max(row["y15"], row["y16"])
        -
        min(row["y0"], row["y1"], row["y2"])
    )

    f["label"] = row["label"]

    features.append(f)

feature_df_yolo = pd.DataFrame(features)

print(feature_df_yolo.shape)

feature_df_yolo.head()
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import StratifiedKFold
from xgboost import XGBClassifier

X = feature_df_yolo.drop(
    columns=["label"]
).values

le = LabelEncoder()

y = le.fit_transform(
    feature_df_yolo["label"]
)

model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    random_state=42,
    n_jobs=-1
)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scores = cross_val_score(
    model,
    X,
    y,
    cv=cv,
    scoring='accuracy'
)

print("Fold Scores:")
print(scores)

print("\nMean Accuracy:")
print(scores.mean())
```

```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input
import numpy as np
import cv2

resnet = ResNet50(
    weights='imagenet',
    include_top=False,
    pooling='avg'
)

def extract_feature(img_path):

    img = cv2.imread(img_path)

    img = cv2.cvtColor(
        img,
        cv2.COLOR_BGR2RGB
    )

    img = cv2.resize(
        img,
        (224,224)
    )

    img = np.expand_dims(
        img,
        axis=0
    )

    img = preprocess_input(img)

    feat = resnet.predict(
        img,
        verbose=0
    )

    return feat.flatten()
```

```python
import os
import numpy as np

base_dir = "/content/dataset"

# Use image_name from the final landmark_df which has 402 samples
image_paths = []

for img_name in landmark_df['image_name']:

    found = False

    # Iterate through class folders to find the image
    for cls in os.listdir(base_dir):

        img_path = os.path.join(
            base_dir,
            cls,
            img_name
        )

        if os.path.exists(img_path):

            image_paths.append(img_path)

            found = True
            break

    if not found:
        print(f"Missing image for deep feature extraction: {img_name}")

deep_features = []

for p in image_paths:

    deep_features.append(
        extract_feature(p)
    )

deep_features = np.array(
    deep_features
)

print(deep_features.shape)
```

```python
from sklearn.decomposition import PCA

pca = PCA(
    n_components=0.95,
    random_state=42
)

deep_pca = pca.fit_transform(
    deep_features
)

print(
    deep_pca.shape
)
```

```python
import os
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array, array_to_img

SOURCE_DIR = "/content/dataset"
TARGET_DIR = "/content/augmented_dataset"

os.makedirs(TARGET_DIR, exist_ok=True)

datagen = ImageDataGenerator(
    rotation_range=25,
    width_shift_range=0.2,
    height_shift_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    brightness_range=[0.8,1.2],
    fill_mode='nearest'
)

for cls in os.listdir(SOURCE_DIR):

    src_class = os.path.join(SOURCE_DIR, cls)

    if not os.path.isdir(src_class):
        continue

    dst_class = os.path.join(TARGET_DIR, cls)

    os.makedirs(dst_class, exist_ok=True)

    for img_name in os.listdir(src_class):

        img_path = os.path.join(src_class, img_name)

        img = load_img(img_path)

        x = img_to_array(img)

        x = x.reshape((1,) + x.shape)

        base_name = img_name.split('.')[0]

        # Save original
        img.save(os.path.join(dst_class, img_name))

        count = 0

        for batch in datagen.flow(
            x,
            batch_size=1,
            save_to_dir=dst_class,
            save_prefix=base_name,
            save_format='jpg'
        ):

            count += 1

            if count >= 5:      # Create 5 extra versions
                break

print("Augmentation Complete")
```

```python
import tensorflow as tf

IMG_SIZE = 224
BATCH_SIZE = 32

train_ds = tf.keras.preprocessing.image_dataset_from_directory(
    "/content/augmented_dataset",
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)

val_ds = tf.keras.preprocessing.image_dataset_from_directory(
    "/content/augmented_dataset",
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)
```

```python
import tensorflow as tf
from tensorflow.keras import layers

base_model = tf.keras.applications.ConvNeXtTiny(
    include_top=False,
    weights='imagenet',
    input_shape=(224,224,3),
    pooling='avg'
)

base_model.trainable = False

inputs = tf.keras.Input(shape=(224,224,3))

x = tf.keras.applications.convnext.preprocess_input(inputs)

x = base_model(x)

x = layers.BatchNormalization()(x)

x = layers.Dense(
    512,
    activation='gelu'
)(x)

x = layers.Dropout(0.5)(x)

outputs = layers.Dense(
    5,
    activation='softmax'
)(x)

model = tf.keras.Model(inputs, outputs)

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
import tensorflow as tf

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor='val_accuracy',
    patience=5,
    restore_best_weights=True
)

reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=3
)

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=8,
    callbacks=[
        early_stop,
        reduce_lr
    ]
)
```

```python
model.save("convnext_best.keras")
```

```python
from sklearn.metrics import classification_report
import numpy as np

y_true = []
y_pred = []

for images, labels in val_ds:

    preds = model.predict(images, verbose=0)

    y_true.extend(labels.numpy())
    y_pred.extend(np.argmax(preds, axis=1))

print(classification_report(y_true, y_pred))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(8,6))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("ConvNeXt Confusion Matrix")

plt.show()
```
